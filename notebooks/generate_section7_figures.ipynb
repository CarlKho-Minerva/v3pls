## Figure 7.4: Per-Class F1 Scores Bar Chart

```python
# Example F1 scores from the multiclass classifier
# These values should come from actual classification_report
classes = ['jump', 'punch', 'turn_left', 'turn_right', 'idle', 'noise']
f1_scores_example = [0.917, 0.833, 0.917, 0.833, 0.917, 0.950]  # Example values

fig, ax = plt.subplots(figsize=(12, 7))
colors = ['green' if f1 >= 0.9 else 'orange' if f1 >= 0.85 else 'red' for f1 in f1_scores_example]
bars = ax.bar(classes, f1_scores_example, color=colors, edgecolor='black', linewidth=1.5)

# Add F1 score labels on bars
for i, (cls, f1) in enumerate(zip(classes, f1_scores_example)):
    ax.text(i, f1 + 0.01, f'{f1:.3f}', ha='center', fontweight='bold', fontsize=11)

# Add horizontal line at 0.8 (usability threshold)
ax.axhline(y=0.8, color='black', linestyle='--', linewidth=2, label='80% Threshold', alpha=0.7)

ax.set_xlabel('Gesture Class', fontsize=13, fontweight='bold')
ax.set_ylabel('F1-Score', fontsize=13, fontweight='bold')
ax.set_title('Per-Class F1 Scores (Multiclass Classifier)', fontsize=16, fontweight='bold')
ax.set_ylim(0, 1.05)
ax.grid(True, axis='y', alpha=0.3)
ax.legend()
plt.xticks(rotation=15)

plt.tight_layout()
plt.savefig('../assignment/figures/figure_7_4_f1_scores.png', dpi=300, bbox_inches='tight')
plt.show()

print("Figure 7.4 saved: Per-class F1 scores")
print("Note: Update with actual values from classification_report")
```
